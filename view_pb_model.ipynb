```python
import tensorflow as tf
from tensorflow.python.framework import ops
from tensorflow.python.framework import importer
```

```python
# in case the model is save standardly

# Load the model
model_path = './FacePoseModel1.pb'

# Load the model using TensorFlow
try:
    model = tf.saved_model.load(model_path)
    print("Model loaded successfully!")

    # List the signatures of the model (entry points)
    print("Model signatures:")
    for key in model.signatures:
        print(f"Signature: {key}")
        print(f"Details: {model.signatures[key]}")
except Exception as e:
    print(f"Error loading model: {e}")
```

```python
# Not save standardly, still be able to check frozen graph

def load_frozen_graph(frozen_graph_filename):
    # Load the protobuf file from the disk and parse it to retrieve the computational graph
    with tf.io.gfile.GFile(frozen_graph_filename, "rb") as f:
        graph_def = tf.compat.v1.GraphDef()
        graph_def.ParseFromString(f.read())

    # Import the graph into the current default Graph
    with tf.Graph().as_default() as graph:
        tf.import_graph_def(graph_def, name="")
    return graph

# Path to the frozen graph .pb file
frozen_model_filename = 'FacePoseModel2.pb'

# Load the frozen graph
try:
    graph = load_frozen_graph(frozen_model_filename)
    print("Model loaded successfully!")

    # Print the operations (nodes) in the graph
    for op in graph.get_operations():
        print(op.name)
except Exception as e:
    print(f"Error loading frozen graph: {e}")
```
